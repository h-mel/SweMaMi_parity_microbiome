```r
grep("kit", names(fp_tp1), value = TRUE)
```

```r
# Find participants where kit2 is missing
fp_pn_kit2_missing <- fp_tp1 |>
  filter(is.na(kit2.faecal_sample.barcode)) |>
  pull(Q1_Personnummer)
fp_pn_kit2_missing
```

```r
sp_pn_kit2_missing <-  sp_tp1 |>
  filter(is.na(kit2.faecal_sample.barcode)) |>
  pull(Q1_Personnummer)
sp_pn_kit2_missing
```

```r
# doubble check 
fp_tp1$kit2.faecal_sample.barcode
sp_tp1$kit2.faecal_sample.barcode
```

```r
# combine both sets of participants to drop (just good practice here) 
combined_missing_kit2 <- c(fp_pn_kit2_missing, sp_pn_kit2_missing)
length(combined_missing_kit2)
```

# -------------------------------------------------------------------------------

```r
# Keep only participants with valid tp2 faecal samples
fp_tp2 <- fp_tp1 |>
  filter(!(Q1_Personnummer %in% combined_missing_kit2))
head(fp_tp2)
dim(fp_tp2)
```

```r
sp_tp2 <- sp_tp1 |>
  filter(!(Q1_Personnummer %in% combined_missing_kit2))
head(sp_tp2)
dim(sp_tp2)
```

```r
# Find barcodes that are missing from valid sample list
missing_fp_tp2_samples <- setdiff(fp_tp2$kit2.faecal_sample.barcode, sample_id)
length(missing_fp_tp2_samples)
```

```r
missing_sp_tp2_samples <- setdiff(sp_tp2$kit2.faecal_sample.barcode, sample_id)
length(missing_sp_tp2_samples)
```

```r
# This part is mostly good practice since no more samples were missing
# Combine all missing samples (good practice)
all_missing_samples <-c(missing_fp_tp2_samples, missing_sp_tp2_samples)
all_missing_samples
```

```r
# Get the personal numbers
fp_pn_to_drop <- fp_tp2 |>
  filter(kit2.faecal_sample.barcode %in% all_missing_samples) |>
  pull(Q1_Personnummer)
length(fp_pn_to_drop)
```

```r
sp_pn_to_drop <- sp_tp2 |>
  filter(kit2.faecal_sample.barcode %in% all_missing_samples) |>
  pull(Q1_Personnummer)
length(sp_pn_to_drop)
```

```r
# Combine all to drop
combined_pn_to_drop <- c(fp_pn_to_drop, sp_pn_to_drop)
length(combined_pn_to_drop)
```

```r
# Keep only participants with valid sample
fp_tp2_filtered_nulliparous <- fp_tp2 |>
  filter(!(Q1_Personnummer %in% combined_pn_to_drop)) 
head(fp_tp2_filtered_nulliparous)
dim(fp_tp2_filtered_nulliparous)
```

```r
sp_tp2_filtered_nulliparous <- sp_tp2 |>
  filter(!(Q1_Personnummer %in% combined_pn_to_drop))
head(sp_tp2_filtered_nulliparous)
dim(sp_tp2_filtered_nulliparous)
```

```r
# Contain only participants with valid, non-missing faecal sample barcodes for kit2 (AND kit1)
# Final check: Make sure both fp and sp have the same participants with intersect
 length(intersect(fp_tp2_filtered_nulliparous$Q1_Personnummer, sp_tp2_filtered_nulliparous$Q1_Personnummer))
```
