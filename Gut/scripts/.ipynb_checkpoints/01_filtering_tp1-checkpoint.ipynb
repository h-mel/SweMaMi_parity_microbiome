```r
library(tidyverse)
library(dplyr)
```

# Reading files~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~

```r
# Read cleaned first and second pregnancy
fp <-  read.csv("/ceph/projects/010_SweMaMi/bm.nowak/unfiltered_divided_data/2025-07-29_kit_fp_.csv", header = TRUE)
dim(fp)
head(fp)
```

```r
sp <-  read.csv("/ceph/projects/010_SweMaMi/bm.nowak/unfiltered_divided_data/2025-07-29_kit_sp_.csv", header = TRUE)
dim(sp)
head(sp)
```

```r
sample_id <- readLines("/ceph/projects/010_SweMaMi/bm.nowak/data/list_of_samples_after_filtration_process.txt")
length(sample_id)
```

```r
# Find participants in fp where fecal barcode is missing
pn_fp_tp1 <- fp |>
  filter(is.na(kit1.faecal_sample.barcode)) |>
  pull(Q1_Personnummer)
pn_fp_tp1
```

```r
# Find participants in sp where faecal barcode is missing
pn_sp_tp1 <- sp |>
  filter(is.na(kit1.faecal_sample.barcode)) |>
  pull(Q1_Personnummer)
pn_sp_tp1
```

```r
# Combine both sets of participants to drop
drop_tp1_pn <- c(pn_fp_tp1, pn_sp_tp1)
drop_tp1_pn
length(unique(drop_tp1_pn))
```

# ===============================================================================

```r
# Keep only participants with valid TP1 faecal samples
fp_tp1 <- fp |>
  filter(!(Q1_Personnummer %in% drop_tp1_pn))
head(fp_tp1)
dim(fp_tp1)
```

```r
sp_tp1 <- sp |> 
  filter(!(Q1_Personnummer %in% drop_tp1_pn))
head(sp_tp1)
dim(sp_tp1)
```

```r
# Find barcodes in fp that are missing from the valid sample list
missing_fp_tp1_samples <- setdiff(fp_tp1$kit1.faecal_sample.barcode, sample_id)
```

```r
# Same for sp
missing_sp_tp1_samples <- setdiff(sp_tp1$kit1.faecal_sample.barcode, sample_id)
```

```r
# Combine all missing sample barcodes
all_missing_samples <-  c(missing_fp_tp1_samples, missing_sp_tp1_samples)
length(all_missing_samples)
```

```r
# Get person num for those missing samples from fp and sp
pn_fp_to_drop <- fp_tp1 |>
  filter(kit1.faecal_sample.barcode %in% all_missing_samples) |>
  pull(Q1_Personnummer)
```

```r
pn_sp_to_drop <-  sp_tp1 |>
  filter(kit1.faecal_sample.barcode %in% all_missing_samples) |>
  pull(Q1_Personnummer)
```

```r
# Combine all personal num to drop
pn_drop_all <- c(pn_fp_to_drop, pn_sp_to_drop)
pn_drop_all
length(unique(pn_drop_all))
```

```r
# Ensure we only retain those whose samples are both present and valid
# Final cleaned dataset after filtering out bad samples
fp_tp1_filt <- fp_tp1 |>
  filter(!(Q1_Personnummer %in% pn_drop_all))
dim(fp_tp1_filt)
```

```r
sp_tp1_filt <- sp_tp1 |>
  filter(!(Q1_Personnummer %in% pn_drop_all))
dim(sp_tp1_filt)
```

```r
# These contain only participants with valid, non-missing faecal sample barcodes also found in sample_id.
# Final check: Make sure both fp and sp have the same participants
# intersect() confirms the alignment.
length(intersect(fp_tp1_filt$Q1_Personnummer, sp_tp1_filt$Q1_Personnummer))
```

```r
# Write to file
write.table(fp_tp1_filt, file = paste0("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/", Sys.Date(), "_fp_tp1_filt", ".csv"), sep = ",", row.names = FALSE)
write.table(sp_tp1_filt, file = paste0("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/", Sys.Date(), "_sp_tp1_filt", ".csv"), sep = ",", row.names = FALSE)
```
