```r
# Read in filtered data 
fp <-  read.csv("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/feacal/tp1_filtered/2025-07-30_fp_tp1_filt.csv", header = TRUE)
dim(fp)
head(fp)
```

```r
sp <- read.csv("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/feacal/tp1_filtered/2025-07-30_sp_tp1_filt.csv", header = TRUE)
head(sp)
dim(sp)
```

```r
# Find fp women where birth date of child is missing
pn_fp_missing_birthdate <- fp |>
  filter(is.na(fodelsedatum_barn)) |>
  pull(Q1_Personnummer)
pn_fp_missing_birthdate
length(unique(pn_fp_missing_birthdate))
```

```r
# Keep only participants where birthdate exists
fp_filtered_birthdate <- fp |>
  filter(!(Q1_Personnummer %in% pn_fp_missing_birthdate)) 
dim(fp_filtered_birthdate)
```

```r
# Find fp women with primipara = TRUE
pn_fp_primipara <- fp |>
  filter(fp$Primipara == 1) |>
  pull(Q1_Personnummer) 
```

```r
# Keep only participants that are primiparous

fp_filtered <- fp_filtered_birthdate |>
  filter(Q1_Personnummer %in% pn_fp_primipara) 
head(fp_filtered)
dim(fp_filtered)
```

```r
# Find all primipara personal# from fp
pn_filtered <- fp_filtered |>
  pull(Q1_Personnummer)
pn_filtered
length(pn_filtered)
```

```r
# Keep sp participants primiparous in fp
sp_filtered <- sp |>
  filter(Q1_Personnummer %in% pn_filtered)
head(sp_filtered)
dim(sp_filtered)
```

```r
# Doubble check that all are no longer primiparous
pn_check <- sp_filtered |> 
  filter(sp_filtered$Primipara == 1) |>
  pull(Q1_Personnummer)
pn_check
length(pn_check)
```

```r
# Double check that birthdates present
pn_check2 <- sp_filtered |>
  filter(is.na(fodelsedatum_barn)) |>
  pull(Q1_Personnummer)
pn_check2
```
