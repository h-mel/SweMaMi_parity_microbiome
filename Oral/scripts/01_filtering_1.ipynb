# Join df -------------------------------------------------------------------------------------------------------------------------

 Files are merged horizontaly (each file added to the right).
- While adding, check that each row matches and if there are new rows then add new row.
- full_join merges by keeping rows from both and adding NA if no match.
- Join by using species name (clade) 

```r
# matches first "s__", then any char after it ("+"),  may end "([...])?" with "_unclassified", 
# match goes to end of str "$"
match_pattern <- "s__.+(_unclassified)?$" 
```

```r
# Reads each file in setwd if file name present in input list
# Also only keeps rows that match pattern
f_join_tables <- function(input_list_of_files){
    lapply(input_list_of_files, function(a_file){
        
        files_tabled <- read.table(a_file, header = TRUE)
        
        #cat("\nProcessing:", a_file, "\n") # debug line
        print(dim(files_tabled)) # Dim for each df
        
        # finds rows where pattern TRUE (excludes unnecessary taxa info etc since pattern match)
        filtered_tab <- files_tabled |> filter(grepl(match_pattern, clade_name)) 
                
        return(filtered_tab)
    })
    }

all_df_list <- f_join_tables(file_list)
head(all_df_list[[1]]) # check first data frame
```

```r
head(all_df_list[[32]]) # check last data frame
```

```r
# full_join merges by keeping rows from both and adding NA if no match and join by using species name (clade) 

joined_table <- reduce(all_df_list, full_join, by = "clade_name") 
# reduce takes each df and joins them together with the previous ones, 
# I.e instead of looping: append this_df with that_df, then append this_df with that_df_2 etc, it takes resulting df and full joins the next df in the list to it

#head(joined_table)
dim(joined_table)
tail(joined_table)
#colnames(joined_table)
```

## Find and Extract highest reads

If the prefix of colnames in 'y_exclude_filt_new_arrange' appears more than once, then check reads for that prefix in 'joined_table_id_cols' --> see which flowcell has more reads (after_kraken2_host_removal)
The flowcell coresponding to higher reads is keept 

```r
# Extract flowcells and sampleids
colnames_info <- colnames(y_exclude_filt_new_arrange)
prefix_info <- str_extract(colnames_info, '^[^_]+')
sort(prefix_info) |> head(prefix_info, n = 20) 

flowcell_ids <- str_extract(colnames_info, '(?<=__)[^x]+')
head(flowcell_ids, n = 20)
```

```r
# Extract duplicate samples
duplicate_ids <- prefix_info[duplicated(prefix_info) | duplicated(prefix_info, fromLast = TRUE)] |> unique()
sort(duplicate_ids) |> head(duplicate_ids, n = 20) 
length(duplicate_ids)
```

```r
# For each duplicate sampleID, find flowcell with the highest reads
duplicate_rows <- joined_table_id_cols |> filter(SampleID %in% duplicate_ids)

dim(duplicate_rows)
#duplicate_rows |> sort_by(duplicate_rows$SampleID) |> head()
```

```r
# Find and extract flowcell with highest reads
highest_reads_flowcell <- duplicate_rows |>
    group_by(SampleID) |>
    slice_max(after_kraken2_host_removal, n = 1, with_ties = FALSE) |> # Inside each sampleID group: selects highest reads
    ungroup() |>
    select(SampleID, EID)

dim(highest_reads_flowcell)
head(highest_reads_flowcell)
```

Maybe this function is what causes issues with the reads?:

```r
# Which columns to keep 
# sapply: collects all (TRUE/FALSE) into logical vector keep_cols 
keep_cols <- sapply(seq_along(colnames_info), function(i){ # loop over all columns 1-n, and for each column do
    
    # If sample (prefix) is not duplicate then keep it
    if (!(prefix_info[i] %in% duplicate_ids)) return(TRUE)

    # otherwise the sample is duplicate; then keep only if flowcell matches best flowcell for that sample  
    best_flowcell <- highest_reads_flowcell$EID[highest_reads_flowcell$SampleID == prefix_info[i]]
    prefix_info[i] %in% best_flowcell
})

best_reads_filt_new_arrange <- y_exclude_filt_new_arrange[, keep_cols]
dim(best_reads_filt_new_arrange)
head(best_reads_filt_new_arrange)
```
