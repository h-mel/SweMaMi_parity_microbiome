# 10 — Oral microbiome parity

**Manuscript:**  Investigating the Effect of Parity on Gut, Oral and Vaginal Microbiome of Pregnant Women 

**Purpose:** Computes observed species richness, Shannon diversity, and Pielou's evenness from MetaPhlAn4 relative abundances for the oral microbiome. Compares nulliparous (First pregnancy) vs primiparous (Second pregnancy) samples at TP1 using paired tests, reflecting the within-subject design. Beta diversity (Bray-Curtis, Jaccard, robust Aitchison) is tested with PERMANOVA using permutations restricted within participant.

**Body site:** Oral

**Design:** Each woman contributes a sample in both pregnancies at TP1: gestational week 10–20;.Parity is a within-subject factor.

**Produces:** Tbale2, TableS3

**Programmer:** Anusha Antony and Barbara Nowak

```r
# sp
sp_even <- sp_shan/log(sp_sr)
sp_even
```

### 3.2 Compare nulliparous vs primiparous at TP1 (paired)

```r
summary(fp_shan)
```

```r
df_shannon <-  data.frame(Shannon = c(fp_shan, sp_shan), Group = rep(c("Nulliparous", "Primiparous"), times = c(length(fp_shan), length(sp_shan))))
head(df_shannon)
class(df_shannon$Shannon)
```

```r
table(df_shannon$Group)
```

```r
df_shannon_1 <- df_shannon |>
  rownames_to_column(var = "Sample_ID")
head(df_shannon_1)
```

```r
table(df_shannon_1$Group)
```

```r
df_shannon_1 <- df_shannon_1 |>
  left_join(
    combined_fp_sp_tp1 |>
      select(
        kit1.salival_sample.barcode ,
        Q1_Personnummer
      ),
    by = c("Sample_ID" = "kit1.salival_sample.barcode")
  )
head(df_shannon_1 )
```

```r
shannon_wide <- df_shannon_1 |>
  select(Q1_Personnummer, Group, Shannon) |>
  pivot_wider(
    names_from = Group,
    values_from = Shannon
    )
shannon_wide |> head()
```

```r
#Normally distributed or not?
shannon_wide$difference <- shannon_wide$Primiparous - shannon_wide$Nulliparous
shapiro.test(shannon_wide$difference)

#p > 0.05 -> paired differences don't significantly deviate from normal -> paired t-test
```

```r
# Paired Wilcoxon signed-rank test
result_shan <- wilcox.test(
  shannon_wide$Nulliparous,
  shannon_wide$Primiparous,
  paired = TRUE,
  exact = FALSE,
  conf.int = TRUE
)

result_shan

# Extract p-value
shan_p <- result_shan$p.value
shan_p
```

```r
df_richness <-  data.frame(Species_richness = c(fp_sr, sp_sr), Group = rep(c("Nulliparous", "Primiparous"), times = c(length(fp_sr), length(sp_sr))))
head(df_richness)
```

```r
# Calculate difference for Shapiro test
df_richness_1 <- df_richness |>
  rownames_to_column(var = "Sample_ID")
head(df_richness_1)
```

```r
# Extract personal numbers for barcodes and add new column with pn to df_richness_1
df_richness_1 <- df_richness_1 |>
  left_join(
    combined_fp_sp_tp1 |>
      select(
        kit1.salival_sample.barcode ,
        Q1_Personnummer
      ),
    by = c("Sample_ID" = "kit1.salival_sample.barcode")
  )
head(df_richness_1 )
```

```r
richness_wide <- df_richness_1 |>
  select(Q1_Personnummer, Group, Species_richness) |>
  pivot_wider(
             names_from = Group,
             values_from = Species_richness
           )
head(richness_wide )
```

```r
# Test data distribution
richness_wide$difference <- richness_wide$Primiparous - richness_wide$Nulliparous
shapiro.test(richness_wide$difference)
```

```r
result_rich <- t.test(
  richness_wide$Nulliparous,
  richness_wide$Primiparous,
  paired = TRUE,
  conf.level = 0.95
)
result_rich
```

```r
df_evenness <-  data.frame(Evenness = c(fp_even, sp_even), Group = rep(c("Nulliparous", "Primiparous"), times = c(length(fp_even), length(sp_even))))
head(df_evenness)
class(df_evenness$Evenness)
```

```r
df_evenness <- df_evenness |>
  rownames_to_column(var = "Sample_ID")
head(df_evenness)
```

```r
 #Extract personal numbers for barcodes and add new column with pn
df_evenness_1 <- df_evenness|>
  left_join(
    combined_fp_sp_tp1 |>
      select(
        kit1.salival_sample.barcode,
        Q1_Personnummer
      ),
    by = c("Sample_ID" = "kit1.salival_sample.barcode")
  )
head(df_evenness_1 )
```

```r
evenness_wide <- df_evenness_1 |>
  select(Q1_Personnummer, Group, Evenness) |>
  pivot_wider(
    names_from = Group,
    values_from = Evenness
  )
head(evenness_wide)
```

```r
evenness_wide$difference <- evenness_wide$Primiparous - evenness_wide$Nulliparous
shapiro.test(evenness_wide$difference)
```

```r
result_evenness <- t.test(
  evenness_wide$Nulliparous,
  evenness_wide$Primiparous,
  paired = TRUE,
  conf.level = 0.95
)

result_evenness
```

## 4. Beta diversity

```r
combined_fp_sp_tp1$Time_point <- "TP1"
```

```r
combined_fp_sp_tp1 <- combined_fp_sp_tp1 %>%
  rename(Sample_id = kit1.salival_sample.barcode)
```

```r
meta_tp1 <- combined_fp_sp_tp1 %>% 
            select(Studienummer,Q1_Personnummer,Sample_id, Group, Time_point)
```

```r
rownames(meta_tp1) <- meta_tp1$Sample_id
head(meta_tp1)
```

```r
taxa_tab_tp1 <- taxa_tab[, colnames(taxa_tab) %in% meta_tp1$Sample_id]

head(taxa_tab_tp1)

dim(taxa_tab_tp1)
```

```r
taxa_tp1 <- as.data.frame(t(taxa_tab_tp1))
class(taxa_tp1)
head(taxa_tp1)
```

```r
ordered_taxa_tp1 <- taxa_tp1[rownames(meta_tp1), ]
head(ordered_taxa_tp1)
```

```r
all(rownames(ordered_taxa_tp1) == rownames(meta_tp1))
```

### 4.1 Calculate distance matrices

```r
dist_bc_tp1 <- vegdist(ordered_taxa_tp1, method = "bray")
dist_jac_tp1 <- vegdist(ordered_taxa_tp1, method = "jaccard", binary = TRUE)
dist_ait_tp1 <- vegdist(ordered_taxa_tp1, method = "robust.aitchison")
```

```r
all(labels(dist_bc_tp1) == rownames(meta_tp1))
all(labels(dist_jac_tp1) == rownames(meta_tp1))
all(labels(dist_ait_tp1) == rownames(meta_tp1))
```

### 4.2. Dispersion check (betadisper)

```r
set.seed(123)
dispersion_J_tp1 <- vegan::betadisper(dist_jac_tp1, meta_tp1$Group)
permutest(dispersion_J_tp1)
```

```r
set.seed(123)
dispersion_tp1_At <- vegan::betadisper(dist_ait_tp1 , meta_tp1$Group)
permutest(dispersion_tp1_At)
```

```r
set.seed(123)
dispersion_tp1_BC <- vegan::betadisper(dist_bc_tp1, meta_tp1$Group)
permutest(dispersion_tp1_BC )
```

### 4.3 PERMANOVA — nulliparous vs primiparous at TP1

```r
table(meta_tp1$Q1_Personnummer, meta_tp1$Group)
```

```r
ctrl_tp1 <- how(plots  = Plots(strata = meta_tp1$Q1_Personnummer, type = "none"),
                within = Within(type = "free"), nperm = 999)

```

```r
set.seed(1234)
tp1_adonis_J <- adonis2(dist_jac_tp1 ~ Q1_Personnummer + Group, data = meta_tp1,
        permutations = ctrl_tp1, by = "terms")
tp1_adonis_J 
r2_j_tp1 <- tp1_adonis_J["Group", "R2"]
p_j_tp1  <- tp1_adonis_J["Group", "Pr(>F)"]
r2_j_tp1
p_j_tp1 
```

```r
set.seed(1234)
tp1_adonis_AT <- adonis2(dist_ait_tp1  ~ Q1_Personnummer + Group, data = meta_tp1,
        permutations = ctrl_tp1, by = "terms")
tp1_adonis_AT 
r2_AT_tp1 <- tp1_adonis_AT["Group", "R2"]
p_AT_tp1  <- tp1_adonis_AT["Group", "Pr(>F)"]
r2_AT_tp1
p_AT_tp1 
```

```r
set.seed(1234)
tp1_adonis_BC <- adonis2(dist_bc_tp1 ~ Q1_Personnummer + Group, data = meta_tp1,
        permutations = ctrl_tp1, by = "terms")
tp1_adonis_BC
r2_BC_tp1 <- tp1_adonis_BC["Group", "R2"]
p_BC_tp1  <- tp1_adonis_BC["Group", "Pr(>F)"]
r2_BC_tp1
p_BC_tp1 
```
