## Kit1-------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
# Find pn where kit1_saiva is missing
pn_fp_tp1 <- fp_saliva |>
  filter(is.na(kit1.salival_sample.barcode)) |>
  pull(Q1_Personnummer)
pn_fp_tp1
length(pn_fp_tp1)
```

```r
pn_sp_tp1 <- sp_saliva |>
    filter(is.na(kit1.salival_sample.barcode)) |>
    pull(Q1_Personnummer)
pn_sp_tp1
length(pn_sp_tp1)
```

```r
# combine participants to drop
drop_pn_combined <- c(pn_fp_tp1, pn_sp_tp1)
unique(drop_pn_combined) |> length() 
```

```r
fp_tp1_saliva <- fp_saliva |>
    filter(!(Q1_Personnummer %in% drop_pn_combined))
head(fp_tp1_saliva)
dim(fp_tp1_saliva)
```

```r
sp_tp1_saliva <- sp_saliva |>
    filter(!(Q1_Personnummer %in% drop_pn_combined))
head(sp_tp1_saliva)
dim(sp_tp1_saliva)
```

## Missing birthweight drop ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
# Find fp women where birth date of child is missing
# Why did we do this? oR why would the weight be missing?
pn_fp_missing_birthdate <- fp_tp1_saliva |>
  filter(is.na(fodelsedatum_barn)) |>
  pull(Q1_Personnummer)
pn_fp_missing_birthdate
```

```r
# Find women where birth weight is missing
grep("weight", names(fp_tp1_saliva), value = TRUE)
pn_fp_missing_birthweight <- fp_tp1_saliva |>
    filter(is.na(Birthweight_checked)) |>
    pull(Q1_Personnummer)
pn_fp_missing_birthweight
```

```r
setdiff(pn_fp_missing_birthdate,pn_fp_missing_birthweight)
```

```r
intersect(pn_fp_missing_birthdate, drop_pn_combined)
```

```r
# Drop participants with missing birthweight
fp_tp1_saliva_adjusted <- fp_tp1_saliva |>
    filter(!(Q1_Personnummer %in% pn_fp_missing_birthweight))
head(fp_tp1_saliva_adjusted)
dim(fp_tp1_saliva_adjusted)
```

## Nulliparous selection--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
# Find all primipara personal# from fp
pn_primipara_fp <- fp_tp1_saliva_adjusted |>
    filter(fp_tp1_saliva_adjusted$Primipara == 1) |>
    pull(Q1_Personnummer)
pn_primipara_fp
pn_fp_missing_birthdate
```

```r
intersect(pn_primipara_fp, pn_fp_missing_birthdate)
```

```r
# Keep only participants that are primiparous
fp_nulliparous <- fp_tp1_saliva_adjusted |>
  filter(Q1_Personnummer %in% pn_primipara_fp) 
head(fp_nulliparous)
dim(fp_nulliparous)
```

```r
pn_filtered <- fp_nulliparous |>
  pull(Q1_Personnummer)
pn_filtered
length(pn_filtered)
```

```r
# Keep sp participants primiparous in fp
sp_nulliparous <- sp_tp1_saliva |>
  filter(Q1_Personnummer %in% pn_filtered)
head(sp_nulliparous)
dim(sp_nulliparous)
```

```r
# Doubble check that all are no longer primiparous in sp
pn_check <- sp_nulliparous |> 
  filter(sp_nulliparous$Primipara == 1) |>
  pull(Q1_Personnummer)
pn_check
length(pn_check)
#sp_nulliparous |> 
 # filter(sp_nulliparous$Primipara == 1) |> pull(Primipara)
```
