```r
library(tidyverse)
library(dplyr)
```

```r
taxa_tab <- read.table("/ceph/projects/010_SweMaMi/bm.nowak/data/2025-12-09_no_minimum_prev_abundance_no_other_taxa_saliva_taxa_cleaned_both_tp.tsv", header = TRUE, sep = "\t", check.names = FALSE )
dim(taxa_tab)
colSums(taxa_tab) |> sort ()
```

```r
head(taxa_tab)
dim(taxa_tab)
```

```r
# Adding other taxa row
new_row <- c()
tot <- colSums(taxa_tab)
for(i in tot){
    if(i < 100){
        new_row <- c(new_row, 100 - i)
    } else if (i >= 100){
        new_row <- c(new_row, 0)
    }
}
head(tot)
head(new_row)

# add this vector as row to the taxa table
new_row <- setNames(new_row, names(taxa_tab))
#new_row
new_row2 <- as.data.frame(t(new_row))
#new_row2
rownames(new_row2) <- "Other_taxa"
best_filt_new <- rbind(taxa_tab ,new_row2) 
#colSums(best_filt_new )
#now the colSums == 100

filt_new_arrange <- as.data.frame(best_filt_new)
head(filt_new_arrange)
dim(filt_new_arrange)
```

```r
tail(filt_new_arrange)
```

```r
colSums(filt_new_arrange) |> sort()
```

```r
# Export as .tsv
write.table(filt_new_arrange, file = paste0("/ceph/projects/010_SweMaMi/bm.nowak/data/", Sys.Date(),
                                       "no_minimum_abundance_other_taxa_saliva_taxa_table_cleaned_both_tp", ".tsv"), sep = "\t")
```
