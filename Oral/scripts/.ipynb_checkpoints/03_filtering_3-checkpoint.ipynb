# Find valid barcodes ------------------------------------------------------------------------------

```r
# Read fp and sp metadata files to obtain barcodes, OBS $Group for FP has a SPACE at the end!!
combined_fp_sp_tp1$Group
fp_bc <- combined_fp_sp_tp1 |>
  filter(Group == "First pregnancy ") |>
  pull(kit1.salival_sample.barcode)
print(fp_bc) |> length()
```

```r
sp_bc <- combined_fp_sp_tp1 |>
  filter(Group == "Second pregnancy") |>
  pull(kit1.salival_sample.barcode)
print(sp_bc) |> length()
```

FP is the one that has less barcodes extract those since participants will be removed form SP if they are missing int FP.  
Then check what pn numbers correspont to those barcodes <-- this is what the filtering will then be based on. 

```r
valid_barcodes_fp <- taxa_tab[, colnames(taxa_tab) %in% fp_bc] |> colnames()

valid_barcodes_fp
length(valid_barcodes_fp)
```

```r
invalid_pn_fp <- fp_tp1 |> filter(!fp_tp1$kit1.salival_sample.barcode %in% valid_barcodes_fp) |> pull(Q1_Personnummer)
length(invalid_pn_fp)
    head(invalid_pn_fp)
```

```r
valid_barcodes_sp <- taxa_tab[, colnames(taxa_tab) %in% sp_bc] |> colnames()

valid_barcodes_sp
length(valid_barcodes_sp)
```

```r
invalid_pn_sp <- sp_tp1 |> filter(!sp_tp1$kit1.salival_sample.barcode %in% valid_barcodes_sp) |> pull(Q1_Personnummer)
length(invalid_pn_sp)
    head(invalid_pn_sp)
```

```r
class(valid_barcodes_fp)
class(valid_barcodes_sp)
```

```r
intersect(invalid_pn_sp, invalid_pn_sp) |> length()
unique(invalid_pn_fp, invalid_pn_sp) |> length()
```

```r
unique(invalid_pn_fp, invalid_pn_sp) |> length()
```

```r
# combine both personal number to drop
pn_to_drop <- c(invalid_pn_fp, invalid_pn_sp)
pn_to_drop |> length()
pn_to_drop |> unique() |> length()
```

```r
fp_tp1_saliva <- fp_tp1 |>
    filter(!(Q1_Personnummer %in% pn_to_drop))
head(fp_tp1_saliva)
dim(fp_tp1_saliva)
```

```r
sp_tp1_saliva <- sp_tp1 |>
    filter(!(Q1_Personnummer %in% pn_to_drop))
head(sp_tp1_saliva)
dim(sp_tp1_saliva)
```
