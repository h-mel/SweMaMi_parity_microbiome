```r
library(tidyverse)
library(dplyr)
```

```r
fp_saliva <- read.csv("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/tp1_nulliparous/2025-10-22fp_tp1_nulliparous_saliva.csv", header = TRUE, row.names = 1)
head(fp_saliva)
dim(fp_saliva)
```

```r
sp_saliva <- read.csv("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/tp1_nulliparous/2025-10-06sp_tp1_nulliparous_saliva.csv", header = TRUE, row.names = 1)
head(sp_saliva)
dim(sp_saliva)
```

```r
# Find missing samples in kit2

pn_fp_tp2 <- fp_saliva |>
    filter(is.na(kit2.salival_sample.barcode)) |>
    pull(Q1_Personnummer)
pn_fp_tp2
length(pn_fp_tp2)
```

```r
pn_sp_tp2 <- sp_saliva |>
    filter(is.na(kit2.salival_sample.barcode)) |>
    pull(Q1_Personnummer)
pn_sp_tp2
length(pn_sp_tp2)
```

```r
sp_saliva$kit2.salival_sample.barcode
```

```r
drop_pn_combined_tp2 <- c(pn_fp_tp2, pn_sp_tp2)
length(drop_pn_combined_tp2)
```

```r
fp_tp2_saliva <- fp_saliva |>
    filter(!(Q1_Personnummer %in% drop_pn_combined_tp2))
head(fp_tp2_saliva)
dim(fp_tp2_saliva)
```

```r
sp_tp2_saliva <- sp_saliva |>
    filter(!(Q1_Personnummer %in% drop_pn_combined_tp2))
head(sp_tp2_saliva)
dim(sp_tp2_saliva)
```

```r
fp_tp2_saliva$Primipara
```

```r
# The participants need to be marked as multiparous for sp  
sp_tp2_saliva$Primipara
```

```r
dim(sp_tp2_saliva)
```

```r
write.table(fp_tp2_saliva, file = paste0("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/tp2_nulliparous/", 
                                  Sys.Date(), "_fp_tp2_nulliparous", ".csv"), sep = ",", row.names = FALSE)

write.table(sp_tp2_saliva, file = paste0("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/tp2_nulliparous/", 
                                  Sys.Date(), "_sp_tp2_nulliparous", ".csv"), sep = ",", row.names = FALSE)
```
