```r
metadata_old <- read.csv("/ceph/projects/010_SweMaMi/anusha.antony/Anu_SweMaMi/swemami_04APRIL2024.csv", header = TRUE, row.names = 1)
head(metadata_old)
dim(metadata_old)
```

```r
# Extract the studynumber that have to be fixed
sn_to_check <- combined_fp_sp |> pull(Studienummer)
length(sn_to_check)
class(sn_to_check)
```

```r
class(metadata_old$Q2_Datum)
head(metadata_old$Q2_Datum)
```

```r
# Extract the dates and study number for the studynumbers that have to be fixed
dates_old_Q2 <- metadata_old |> 
    filter(Studienummer %in% sn_to_check) |>
    select(Studienummer, Q2_Datum, kit2.salival_sample.datetime_taken)
head(dates_old_Q2)
dim(dates_old_Q2)
```

```r
# Extract the dates for the studynumbers that have to be fixed
dates_old_Q3 <- metadata_old |> 
    filter(Studienummer %in% sn_to_check) |>
    select(Studienummer, Q3_Datum, kit3.salival_sample.datetime_taken)
head(dates_old_Q3)
dim(dates_old_Q3)
```

```r
# Convert dates into date format and remove time
# This is where the dates missmatch so sort both df with old dates first so they are in the same order as combined_fp_sp
dates_old_Q2_sorted <- dates_old_Q2[
  match(combined_fp_sp$Studienummer, dates_old_Q2$Studienummer),
  ]

#(dates_old_Q2_sorted)

dates_old_Q3_sorted <- dates_old_Q3[
  match(combined_fp_sp$Studienummer, dates_old_Q3$Studienummer),
  ]
#head(dates_old_Q3_sorted)
```

```r
combined_fp_sp$Q2_date_fixed <- as.Date(substr(dates_old_Q2_sorted$Q2_Datum, 1, nchar(dates_old_Q2_sorted$Q2_Datum) - 9), format = "%Y-%m-%d")
combined_fp_sp$Q2_date_fixed
class(combined_fp_sp$Q2_date_fixed)
```

```r
combined_fp_sp$Studienummer
```

```r
# Convert dates into date format and remove time
combined_fp_sp$Q3_date_fixed <- as.Date(substr(dates_old_Q3_sorted$Q3_Datum, 1, nchar(dates_old_Q3_sorted$Q3_Datum) - 9), format = "%Y-%m-%d")
combined_fp_sp$Q3_date_fixed
class(combined_fp_sp$Q3_date_fixed)
```

```r
is.na(combined_fp_sp$Q3_date_fixed) |> sum()
```

```r
# Check that dates seem correct manually

head(combined_fp_sp$Q2_date_fixed)
head(combined_fp_sp$Studienummer)

metadata_old |> filter(Studienummer == "Scf5931f") |> select(Q2_Datum) 
metadata_old |> filter(Studienummer == "S14e664f") |> select(Q2_Datum) 
metadata_old |> filter(Studienummer == "S194e21b") |> select(Q2_Datum) 
```

```r
# Check for missmatches

mismatches <- combined_fp_sp |>
  select(Studienummer, Q2_date_fixed) |>
  left_join(
    metadata_old |> select(Studienummer, Q2_Datum),
    by = "Studienummer",
    suffix = c("_fp_sp", "_old")
  ) |>
  filter(Q2_date_fixed != Q2_Datum)
mismatches
```

```r
# Df for manual check missmatches
date_check_old <- metadata_old |> filter(Studienummer %in% sn_to_check) |> select(Studienummer, Q2_Datum, kit2.salival_sample.datetime_taken, Q3_Datum, kit3.salival_sample.datetime_taken)
dim(date_check_old)
#date_check_old
```

```r
date_check_combined <- combined_fp_sp |> filter(Studienummer %in% sn_to_check) |> select(Studienummer, Q2_date_fixed, Q3_date_fixed )
dim(date_check_combined)
#date_check_combined
```

```r
# Sort date_check_combined based on studienummer order in date_check_combined
date_check_combined_sorted <- date_check_combined[
  match(date_check_old$Studienummer, date_check_combined$Studienummer),
  ]
```

```r
# Create df for manual check of dates
merged_date_check <- cbind(date_check_old, date_check_combined_sorted)
head(merged_date_check)
```

```r
# Extract additional variables of interest and add them to a new data frame
combined_fixed_date <- date_check_combined_sorted
```

```r
combined_variables_sorted <- combined_variables[
  match(date_check_combined_sorted$Studienummer, combined_variables$Studienummer),
  ]
head(combined_variables_sorted)
```

```r
# Create new data frame with all variables of interest

combined_date <- left_join(date_check_combined_sorted, combined_variables_sorted, by = "Studienummer")
#combined_date <- cbind(combined_variables_sorted, date_check_combined_sorted) |> select(Studienummer, Q1_Personnummer, Q1_Datum,kit1.salival_sample.datetime_taken, Q2_date_fixed, 
 #                                                                            kit2.salival_sample.datetime_taken, Q3_date_fixed, kit3.salival_sample.datetime_taken)
head(combined_date)
```

```r
# Since dates were also missing from the kits add them aswell from the sorted merged data frame
combined_date$kit2.salival_sample.datetime_taken <- merged_date_check$kit2.salival_sample.datetime_taken
combined_date$kit3.salival_sample.datetime_taken <- merged_date_check$kit3.salival_sample.datetime_taken
head(combined_date)
```

```r
# Sort the columns
combined_date <- combined_date |> select(Studienummer, Q1_Personnummer, Q1_Datum, kit1.salival_sample.datetime_taken, Q2_date_fixed, kit2.salival_sample.datetime_taken, Q3_date_fixed, kit3.salival_sample.datetime_taken)
head(combined_date)
```

## Calculating difference ----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
# Calculate difference 
```

```r
combined_date$Q1_Datum <- as.Date(combined_date$Q1_Datum)
combined_date$kit1.salival_sample.datetime_taken <- as.Date(combined_date$kit1.salival_sample.datetime_taken)
combined_date$kit2.salival_sample.datetime_taken <- as.Date(combined_date$kit2.salival_sample.datetime_taken)
combined_date$kit3.salival_sample.datetime_taken <- as.Date(combined_date$kit3.salival_sample.datetime_taken)
head(combined_date)
```

```r
combined_date$diff_q1_kit1 <- as.integer(difftime(combined_date$kit1.salival_sample.datetime_taken, combined_date$Q1_Datum, units ="days"))
combined_date$diff_q2_kit2 <- as.integer(difftime(combined_date$kit2.salival_sample.datetime_taken, combined_date$Q2_date_fixed, units ="days"))
combined_date$diff_q3_kit3 <- as.integer(difftime(combined_date$kit3.salival_sample.datetime_taken, combined_date$Q3_date_fixed, units ="days"))

head(combined_date)
```
