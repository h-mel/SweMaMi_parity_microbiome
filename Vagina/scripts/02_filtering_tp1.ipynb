# Reading files~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~

```r
# Read cleaned first and second pregnancy
fp <-  read.csv("/ceph/projects/010_SweMaMi/bm.nowak/unfiltered_divided_data/2025-07-29_kit_fp_.csv", header = TRUE)
dim(fp)
head(fp)
```

```r
sp <-  read.csv("/ceph/projects/010_SweMaMi/bm.nowak/unfiltered_divided_data/2025-07-29_kit_sp_.csv", header = TRUE)
dim(sp)
head(sp)
```

```r
sample_id <- readLines("/ceph/projects/010_SweMaMi/bm.nowak/data/2025-11-06_valid_barcodes_vaginal_nuliparous.txt")
length(sample_id)
```

```r
grep("kit", names(fp), value = TRUE)
```

```r
# Find participants in fp where vaginal barcode is missing
pn_fp_tp1 <- fp |>
  filter(is.na(kit1.vaginal_sample.barcode)) |>
  pull(Q1_Personnummer)
pn_fp_tp1
```

```r
# Find participants in sp where vaginal barcode is missing
pn_sp_tp1 <- sp |>
  filter(is.na(kit2.vaginal_sample.barcode)) |>
  pull(Q1_Personnummer)
pn_sp_tp1
```

```r
# Combine both sets of participants to drop
drop_tp1_pn <- c(pn_fp_tp1, pn_sp_tp1)
drop_tp1_pn
length(drop_tp1_pn)
```

# ===============================================================================

```r
# Keep only participants with valid TP1 faecal samples
fp_tp1 <- fp |>
  filter(!(Q1_Personnummer %in% drop_tp1_pn))
head(fp_tp1)
dim(fp_tp1)
```

```r
sp_tp1 <- sp |> 
  filter(!(Q1_Personnummer %in% drop_tp1_pn))
head(sp_tp1)
dim(sp_tp1)
```

```r
# Find barcodes in fp that are missing from the valid sample list
missing_fp_tp1_samples <- setdiff(fp_tp1$kit1.vaginal_sample.barcode, sample_id)
```

```r
# Same for sp
missing_sp_tp1_samples <- setdiff(sp_tp1$kit1.vaginal_sample.barcode, sample_id)
```

```r
# Combine all missing sample barcodes
all_missing_samples <-  c(missing_fp_tp1_samples, missing_sp_tp1_samples)
length(all_missing_samples)
```

```r
# Get person num for those missing samples from fp and sp
pn_fp_to_drop <- fp_tp1 |>
  filter(kit1.vaginal_sample.barcode %in% all_missing_samples) |>
  pull(Q1_Personnummer)
length(pn_fp_to_drop)
```

```r
pn_sp_to_drop <-  sp_tp1 |>
  filter(kit1.vaginal_sample.barcode %in% all_missing_samples) |>
  pull(Q1_Personnummer)
length(pn_sp_to_drop)
```

```r
# Combine all personal num to drop
pn_drop_all <- c(pn_fp_to_drop, pn_sp_to_drop)
pn_drop_all
length(unique(pn_drop_all))
```

```r
# Ensure we only retain those whose samples are both present and valid
# Final cleaned dataset after filtering out bad samples
fp_tp1_filt <- fp_tp1 |>
  filter(!(Q1_Personnummer %in% pn_drop_all))
dim(fp_tp1_filt)
```

```r
sp_tp1_filt <- sp_tp1 |>
  filter(!(Q1_Personnummer %in% pn_drop_all))
dim(sp_tp1_filt)
```

```r
# These contain only participants with valid, non-missing faecal sample barcodes also found in sample_id.
# Final check: Make sure both fp and sp have the same participants
# intersect() confirms the alignment.
length(intersect(fp_tp1_filt$Q1_Personnummer, sp_tp1_filt$Q1_Personnummer))
```

```r
# make sure all in FP are nulliparous and all SP are NOT nulliparous
fp_primipara_check <- fp_tp1_filt |> 
    filter(fp_tp1_filt$Primipara == 1) |>
    pull(Q1_Personnummer)
length(fp_primipara_check) #REMEMBER to remove those pn that are NOT NULLIPAROUS here!!
```

```r
sp_primipara_check <- sp_tp1_filt |> 
    filter(sp_tp1_filt$Primipara == 1) |>
    pull(Q1_Personnummer)
```

```r
# Extract non primiparous pn in FP
non_primipara <- fp_tp1_filt |> 
    filter(!(fp_tp1_filt$Primipara == 1)) |>
    pull(Q1_Personnummer)
length(non_primipara)
```

```r
int_primipara <- intersect(fp_primipara_check, sp_primipara_check)
int_primipara
```

```r
# check dates of intersecting primipara personal#
fp_tp1_filt |> filter(fp_tp1_filt$Q1_Personnummer %in% int_primipara) |> 
    pull(Q1_Datum, missfall.IUFD_comments ) 

sp_tp1_filt |> filter(sp_tp1_filt$Q1_Personnummer %in% int_primipara) |> 
    pull(Q1_Datum, Q3_X7_Birthdate)
```

```r
grep("date", names(fp_tp1_filt), value = TRUE)
```

```r
fp_tp1_filt |> filter(fp_tp1_filt$Q1_Personnummer %in% int_primipara) |> 
    pull(Q1_Datum, Q1_Personnummer ) 

sp_tp1_filt |> filter(sp_tp1_filt$Q1_Personnummer %in% int_primipara) |> 
    pull(Q1_Datum, Q1_Personnummer)
```

```r
# Combine non primiparous personal# and the 5 participants to drop:
non_primipara_pn_to_drop <- c(non_primipara, int_primipara)
length(non_primipara_pn_to_drop)
```

```r
fp_tp1_nulliparous <- fp_tp1_filt |> 
    filter(!(fp_tp1_filt$Q1_Personnummer %in% non_primipara_pn_to_drop))
dim(fp_tp1_nulliparous)
head(fp_tp1_nulliparous)
```

```r
sp_tp1_nulliparous <- sp_tp1_filt |> 
    filter(!(sp_tp1_filt$Q1_Personnummer %in% non_primipara_pn_to_drop))
dim(sp_tp1_nulliparous)
head(sp_tp1_nulliparous)
```

```r
sp_tp1_nulliparous |> 
    filter(sp_tp1_nulliparous$Primipara == 0) |>
    pull(Q1_Personnummer) |> 
    length()
```
