# Filtering taxa ------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
# Function for removing minimum abundance and prevalence taxa
is_here <- function(invector, min_abund, min_pres){
  min_count <- length(invector) * min_pres
  these <- length(which(invector > min_abund))
  if(these > min_count){
    return(TRUE)
  }
  else{
    return(FALSE)
  }
}

keep <- apply(taxa_tab , MARGIN = 1, is_here, min_abund = 0.001, min_pres = 0.05)
filt_taxa <- taxa_tab [keep,]
head(filt_taxa)
dim(filt_taxa) 
```

```r
colSums(filt_taxa)
#which(colSums(filt_taxa) > 98)
```

```r
#((colSums(filt_taxa) > 80)) 
col_names_to_drop <- names(filt_taxa)[colSums(filt_taxa) < 80]
length(col_names_to_drop)
head(col_names_to_drop)
```

```r
sum(filt_taxa$FF00653393)
sum(filt_taxa$not_sequenced_010SweMaMi_tra16_FF00530583)
sum(filt_taxa$FF00652234)
```

```r
colsum_filt_taxa <- filt_taxa |> 
    select(!all_of(col_names_to_drop))
length(colsum_filt_taxa)
dim(colsum_filt_taxa)
```

```r
new_row <- c()
tot <- colSums(colsum_filt_taxa)
for(i in tot){
    if(i < 100){
        new_row <- c(new_row, 100 - i)
    } else if (i >= 100){
        new_row <- c(new_row, 0)
    }
}
head(tot)
head(new_row)
```

```r
# add this vector as row to the taxa table
new_row <- setNames(new_row, names(colsum_filt_taxa))
#new_row
new_row2 <- as.data.frame(t(new_row))
#new_row2
rownames(new_row2) <- "Other_taxa"
best_filt_new <- rbind(colsum_filt_taxa ,new_row2) 
#colSums(best_filt_new )
#now the colSums == 100

filt_new_arrange <- as.data.frame(best_filt_new)
head(filt_new_arrange)
dim(filt_new_arrange)
```

```r
tail(filt_new_arrange)
dim(filt_new_arrange)
```

```r
barcodes_vaginal <- colnames(filt_new_arrange)
head(barcodes_vaginal)
length(barcodes_vaginal)
```

```r
read_count <- read.table("/ceph/projects/010_SweMaMi/anusha.antony/Anu_SweMaMi/dnbseq_sample_202309180813.csv", sep = ",", header = TRUE)
dim(read_count)
head (read_count)
```

```r
barcodes_in_reads <- read_count |>
    filter(sample_id %in% barcodes_vaginal)
dim(barcodes_in_reads)
length(unique(barcodes_in_reads$sample_id))
```

```r
# Keep only barcodes where the reads are qual to or more than 100k
read_filtered_barcodes <- barcodes_in_reads |> filter(reads_post_hr >= 100000) |> pull(sample_id)
head(read_filtered_barcodes)
length(read_filtered_barcodes)
```
